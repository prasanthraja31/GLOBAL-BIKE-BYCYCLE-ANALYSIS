# RFM calculation

```python
orders["order_date"] = pd.to_datetime(orders["order_date"],format="%d-%m-%Y")
```

```python
analysis_date = orders["order_date"].max() + pd.Timedelta(days=1)
analysis_date
```

```python
rfm=pd.merge(orders,order_items,on='Order_id',how='right')
rfm
```

```python
rfm = rfm.groupby("Customer_id").agg(
    Recency=("order_date", lambda x: (analysis_date - x.max()).days),
    Frequency=("Order_id", "nunique"),
    Monetary=("Total_price", "sum")
).reset_index()
```

```python
rfm
```

```python
rfm["R_Score"] = pd.qcut(rfm["Recency"], 4, labels=[4,3,2,1])
rfm["F_Score"] = pd.qcut(rfm["Frequency"].rank(method="first"),4,labels=[1,2,3,4])
rfm["M_Score"] = pd.qcut(rfm["Monetary"].rank(method="first"),4,labels=[1,2,3,4])
```

```python
rfm["RFM_Score"] = (
    rfm["R_Score"].astype(int)
    + rfm["F_Score"].astype(int)
    + rfm["M_Score"].astype(int))
```

```python
rfm
```

```python
def segment_customer(score):
    if score >= 10:
        return "Champions"
    elif score >= 8:
        return "Loyal Customers"
    elif score >= 6:
        return "Potential Loyalists"
    elif score >= 4:
        return "At Risk"
    else:
        return "Lost Customers"
rfm["Segment"] = rfm["RFM_Score"].apply(segment_customer)        
```

```python
rfm
```

```python
customer_segments = rfm[["Customer_id","Recency","Frequency","Monetary","Segment"]]
```

```python
customer_segments
```

```python
customer_segments.to_sql(
    "customer_segments",
    con=sql_link,
    if_exists="replace",
    index=False)
```
